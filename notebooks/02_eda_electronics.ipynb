# EDA — Electronics υποσύνολο του ESCI

Το ESCI δεν έχει κατηγορία προϊόντος. Για να βρούμε ποια products είναι electronics,
τα αντιστοιχίζουμε με το **meta_Electronics** του Amazon Reviews 2023 (McAuley Lab):

```
ESCI product_id (ASIN)  ──join──  meta_Electronics parent_asin
```

Ένα product θεωρείται electronics αν το ASIN του υπάρχει στο meta_Electronics.
Όλοι οι υπολογισμοί γίνονται από το `src/electronics.py`
(`python -m src.electronics` γράφει τα ίδια στο `results/eda/electronics/`).

**Προαπαιτούμενα:** `python -m src.data` (splits) και `python -m src.meta` (meta parquet).

````python
import sys
from pathlib import Path

ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))

import pandas as pd
from IPython.display import Image, display

from src import electronics as el
from src.data import load_config
from src.eda import style

pd.set_option("display.max_colwidth", 90)
cfg = load_config("configs/data.yaml")
OUT = ROOT / "results/eda/electronics"
OUT.mkdir(parents=True, exist_ok=True)
style()
````

## 1. Τα δύο datasets

````python
esci, meta = el.load(cfg)
print("ESCI small-us:", esci.shape, "| queries:", esci["query_id"].nunique(), "| products:", esci["product_id"].nunique())
print("meta_Electronics:", meta.shape)
meta.head()
````

## 2. Coverage του join

Πόσα ESCI products βρίσκονται στο meta_Electronics, και πόσα queries «περνούν»
με διάφορους ορισμούς του «electronics query».

````python
m, per_q, cov = el.coverage(esci, meta)
display(pd.Series({k: v for k, v in cov.items() if not isinstance(v, dict)}))
display(pd.DataFrame(cov["queries_by_threshold"]).T)
````

````python
el.fig_frac(per_q, OUT)
display(Image(OUT / "electronics_fraction_per_query.png"))
````

## 3. Sanity check: πόσα electronics χάνει το join;

Παίρνουμε queries που είναι **αναμφίβολα** electronics (λέξεις-κλειδιά όπως laptop, hdmi,
headphones) και μετράμε τι ποσοστό των products τους βρίσκει το join. Αν το φίλτρο
δούλευε τέλεια, τα E products αυτών των queries θα έβγαιναν σχεδόν όλα electronics.

````python
kw, kw_missed = el.keyword_check(m)
print("queries με λέξη-κλειδί:", kw["queries"])
display(pd.Series(kw["matched_share_by_label"], name="matched share"))
print("Δείγμα E products που ΔΕΝ βρέθηκαν στο meta:")
kw_missed.head(15)
````

**Εύρημα:** μόνο ~30% των E products σε προφανώς-electronics queries βρίσκονται στο
meta_Electronics — ενώ στη λίστα που λείπουν υπάρχουν καθαρά electronics (SSD, routers, TV).
Το φίλτρο μέσω ASIN έχει **υψηλή ακρίβεια αλλά χαμηλή κάλυψη (recall)**.

Πιθανές αιτίες (δεν έχουν επαληθευτεί ακόμα):
- το ESCI έχει ASIN *παραλλαγών* (child), ενώ το Amazon 2023 κρατά μόνο `parent_asin`·
- προϊόντα του ESCI (2020–21) που δεν υπάρχουν στο dataset του 2023·
- προϊόντα σε άλλη κατηγορία του Amazon 2023 (π.χ. Cell Phones & Accessories).

Άρα το υποσύνολο παρακάτω είναι **προκαταρκτικό**: είναι πραγματικά electronics, αλλά λείπουν πολλά.

## 4. EDA στο υποσύνολο electronics

Ορισμός: query = electronics αν **≥50% των E products** του βρίσκονται στο meta_Electronics
(το E είναι αυτό που ο χρήστης ζητούσε, άρα δείχνει την πρόθεση του query).

````python
sub, elec_q, words, eda_out = el.elec_eda(m, per_q)
display(pd.Series({k: eda_out[k] for k in ["definition", "queries", "rows", "matched_products"]}))
display(pd.DataFrame({k: eda_out[k] for k in
        ["label_dist", "queries_with_C", "candidates_per_query_median", "query_words_median"]}))
````

````python
el.fig_label_compare(eda_out, OUT)
display(Image(OUT / "labels_electronics_vs_all.png"))
````

Στα electronics υπάρχουν περισσότερα **S** και **C** (αξεσουάρ, συμβατά προϊόντα) και λιγότερα I.

````python
display(pd.DataFrame(eda_out["source_dist"]).fillna(0))
el.fig_query_length(words, elec_q, OUT)
display(Image(OUT / "electronics_query_length.png"))
````

## 5. Συμπεράσματα

1. Το join μέσω ASIN βρίσκει μόνο μικρό μέρος του ESCI (βλ. coverage) και, σε προφανώς-electronics
   queries, χάνει ~70% των products → **το υποσύνολο είναι μικρό και ελλιπές**.
2. Με τον ορισμό «≥50% των E» προκύπτουν λίγες εκατοντάδες queries — λίγα για fine-tuning μόνο
   σε αυτά, αρκετά ως **ξεχωριστό test set** αξιολόγησης.
3. Τα electronics queries διαφέρουν: περισσότερα S και C (αξεσουάρ/συμβατότητα), λιγότερα I.
4. Τα meta δίνουν πληροφορία που λείπει από το ESCI (κατηγορίες, τιμή για ~57% των products,
   βαθμολογίες) — χρήσιμη για το RAG.

**Ανοιχτό:** βελτίωση του φίλτρου (child→parent ASIN από τα reviews, ή classifier σε title)
και τελικός ορισμός του electronics set.
