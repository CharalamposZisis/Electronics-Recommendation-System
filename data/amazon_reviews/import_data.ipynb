# Amazon Reviews 2023 — δείγμα reviews

Διερευνητικό: διαβάζει τα πρώτα 1.000 reviews της κατηγορίας Electronics (streaming) για να δούμε τα πεδία.
Τα **metadata** ανά κατηγορία κατεβαίνουν αναπαράξιμα με `python -m src.meta` (βλ. README).

```python
import datasets
datasets.logging.set_verbosity_error()
```

```python
import pandas as pd

url = (
    "https://mcauleylab.ucsd.edu/public_datasets/"
    "data/amazon_2023/raw/review_categories/"
    "Electronics.jsonl.gz"
)

# Διαβάζουμε τα πρώτα 1.000 reviews (ένα chunk), μόνο για να δούμε τη μορφή
reader = pd.read_json(
    url,
    lines=True,
    compression="gzip",
    chunksize=1000
)

df = next(reader)

print(df.shape)
print(df.head())

df.to_csv("electronics_reviews.csv", index=False)
```
